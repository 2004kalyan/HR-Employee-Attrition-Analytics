```python
import pandas as pd

df = pd.read_csv("HR_Employee_Attrition_Cleaned.csv")
```

```python
df.groupby("Attrition")["MonthlyIncome"].agg(
    ["count", "mean", "median", "min", "max"]
).round(2)
```

```python
role_income_attrition = (
    df.groupby("JobRole")
      .agg(
          Employees=("EmployeeNumber", "count"),
          AvgIncome=("MonthlyIncome", "mean"),
          AttritionRate=("Attrition", lambda x: (x == "Yes").mean() * 100)
      )
      .round(2)
)

print(role_income_attrition)
```
